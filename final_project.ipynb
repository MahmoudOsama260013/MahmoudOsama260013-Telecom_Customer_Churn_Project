# Task 4: Object-Oriented Programming

```python
class InvalidMonthlyChargeError(Exception):
    """Raised when monthly charges are negative."""    # from task 5 
    pass

from abc import ABC,abstractmethod
class Customer(ABC):
    def __init__(self, customer_id, tenure, monthly_charges):
        self.customer_id = customer_id
        self.tenure = tenure
        self.monthly_charges = monthly_charges
    @property
    def monthly_charges(self):
        return self._monthly_charges
    @monthly_charges.setter
    def monthly_charges(self, new_charge):
        if new_charge >= 0:
            self._monthly_charges = new_charge
        else:
            raise InvalidMonthlyChargeError("Monthly charges cannot be negative.")   # from task 5

    @abstractmethod
    def display_status(self):
        pass
    
    @abstractmethod
    def is_churn(self):
        pass

class ActiveCustomer(Customer):
    def is_churn(self):
        return False

    def display_status(self):
        return f"Customer {self.customer_id} is Active"

class ChurnCustomer(Customer):
    def is_churn(self):
        return True
    
    def display_status(self):
        return f"Customer {self.customer_id} is Churn"
    

mahmoud = ActiveCustomer(123 , 3 , 50)
print(mahmoud.is_churn())
print(mahmoud.display_status())
try:
    mahmoud.monthly_charges = -1
except Exception as e:
    print(e)
```

# Task 7: Pandas Data Cleaning


#### save the original data without changes

```python
df.head()
```

```python
clean_df = df.copy() 
```

#### clean columns names

```python
clean_df.columns = clean_df.columns.str.strip().str.lower()
print(df.columns)
```

#### clean and replace descriptive names 

```python
clean_df = clean_df.rename(
    columns={
        "customerid": "customer_id",
        "seniorcitizen": "senior_citizen",
        "phoneservice": "phone_service",
        "multiplelines": "multiple_lines",
        "internetservice": "internet_service",
        "onlinesecurity": "online_security",
        "onlinebackup": "online_backup",
        "deviceprotection": "device_protection",
        "techsupport": "tech_support",
        "streamingtv": "streaming_tv",
        "streamingmovies": "streaming_movies",
        "paperlessbilling": "paperless_billing",
        "paymentmethod": "payment_method",
        "monthlycharges": "monthly_charges",
        "totalcharges": "total_charges",
        "partner"     : "is_marred"    # descriptive
    }
)
print(clean_df.columns)
print(clean_df["total_charges"].dtype)
```

#### Missing Values: Identify and handle: drop, fill, or impute

```python
# since total_charges is str then I need to convert its type before

clean_df["total_charges"] = pd.to_numeric(clean_df["total_charges"], errors="coerce")
print(clean_df.duplicated().sum())
print(clean_df["total_charges"].dtype)

# 4. Identify missing values
print("\nMissing values before handling:")
print(clean_df.isnull().sum())



# 5. Handle missing values

# Fill missing total_charges values with the median.

total_charges_median = clean_df["total_charges"].median()

clean_df["total_charges"] = clean_df["total_charges"].fillna(total_charges_median)

print("\nMissing values after handling:")
print(clean_df.isnull().sum())
```

#### Duplicates: Detect and remove duplicate rows

```python

# 6. Detect and remove duplicate rows

duplicate_count = clean_df.duplicated().sum()

print("\nNumber of duplicate rows before cleaning:")
print(duplicate_count)

# Remove duplicated rows to avoid repeated customer records.
clean_df = clean_df.drop_duplicates()

print("\nNumber of duplicate rows after cleaning:")
print(clean_df.duplicated().sum())

```

####  Convert data types


```python
# Convert total_charges to numeric
clean_df["total_charges"] = pd.to_numeric(
    clean_df["total_charges"],
    errors="coerce"
)

# Convert Yes/No columns to boolean
boolean_columns = [
    "is_marred",
    "dependents",
    "phone_service",
    "paperless_billing",
    "churn"
]

for column in boolean_columns:
    clean_df[column] = clean_df[column].map({"Yes": True,"No": False})

# Convert 0/1 column to boolean
clean_df["senior_citizen"] = clean_df["senior_citizen"].astype(bool)

# Convert categorical columns
category_columns = [
    "gender",
    "multiple_lines",
    "internet_service",
    "online_security",
    "online_backup",
    "device_protection",
    "tech_support",
    "streaming_tv",
    "streaming_movies",
    "contract",
    "payment_method"
]

for column in category_columns:
    clean_df[column] = clean_df[column].astype("category")

```

#### Consistency: Standardize text values (e.g., lowercase, trim whitespace)

```python
# Select all text columns.
text_columns = clean_df.select_dtypes(include="str").columns

# Remove extra spaces and convert text to lowercase.
for column in text_columns:
    clean_df[column] = clean_df[column].str.strip().str.lower()
```

#### Outliers: Note any extreme values (optional: handle them)

```python

# Detect possible outliers

# Use the IQR method to identify possible outliers
# in the monthly_charges column.
q1 = clean_df["monthly_charges"].quantile(0.25)
q3 = clean_df["monthly_charges"].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

monthly_charge_outliers = clean_df[(clean_df["monthly_charges"] < lower_bound)|(clean_df["monthly_charges"] > upper_bound)]

print("\nNumber of possible monthly charge outliers:")
print(len(monthly_charge_outliers))

# The outliers are not removed because high monthly charges
# may represent valid premium service plans rather than incorrect data.
```

```python
# Final verification

print("\nCleaned dataset shape:")
print(clean_df.shape)

print("\nCleaned dataset data types:")
print(clean_df.dtypes)

print("\nFirst five rows of the cleaned dataset:")
print(clean_df.head())
```

# Task 8: Data Analysis

### Which customers have high monthly charges?
### Which contract type has the highest average monthly charges?
### How many customers churned and how many remained?
### How many customers use each payment method?

```python
# Filtering
high_charge = clean_df[
    clean_df["monthly_charges"] > 80
]

print(high_charge.head())
print("Number of customers with hight charges:", len(high_charge))
#This analysis filters customers with monthly charges greater than 80. 
# These customers represent the highest-paying group and may require further analysis to determine whether they are more likely to churn.
#  Understanding this group can help the company improve customer retention.
```

```python
#Grouping + Aggregation + Sorting
# that show the mean charges for every type of contract
contract_analysis = (
    clean_df.groupby("contract")["monthly_charges"]
    .mean()
    .sort_values(ascending=False)
)

print(contract_analysis)
#This analysis compares the average monthly charges for each contract type. 
# Sorting the results makes it easy to identify which contract has the highest average charges.
#  These insights can help the company understand customer spending patterns across different contracts.
```

```python
number_of_churn = clean_df["churn"].sum()
number_of_active = len(clean_df) - number_of_churn

print(f"Number of churn customers: {number_of_churn}")
print(f"Number of active customers: {number_of_active}")
#This analysis shows the number of customers who stayed with the company and those who left.
#  Comparing these two groups helps evaluate customer retention and the overall churn rate.
#  The results can support future strategies to reduce customer loss.
```

```python
payment_count = (
    clean_df.groupby("payment_method")["customer_id"]
    .count()
    .sort_values(ascending=False)
)

print(payment_count)
# This analysis counts the number of customers using each payment method.
#  It helps identify the most commonly used payment option. 
# These results can support decisions related to billing and payment services.
```

# Task 9: Excel Export

```python

# Summary sheet
summary = pd.DataFrame({
    "Finding": [
        "High charge customers",
        "Contract with highest average monthly charges",
        "Most common payment method"    ],
    "Result": [
        len(high_charge),
        contract_analysis.idxmax(),
        payment_count.idxmax()
    ]
})

# Export to Excel
with pd.ExcelWriter("Project_Report.xlsx", engine="openpyxl") as writer:
    clean_df.to_excel(writer, sheet_name="Cleaned Data", index=False)

    contract_analysis.to_frame(
        name="Average Monthly Charges"
    ).to_excel(writer, sheet_name="Analysis", startrow=0)

    payment_count.to_frame(
        name="Customer Count"
    ).to_excel(writer, sheet_name="Analysis", startrow=6)

    summary.to_excel(writer, sheet_name="Summary", index=False)

print("Project_Report.xlsx created successfully!")
```

```python
from openpyxl import load_workbook
from openpyxl.styles import Font

wb = load_workbook("Project_Report.xlsx")

for sheet in wb.sheetnames:
    ws = wb[sheet]

    # Bold headers
    for cell in ws[1]:
        cell.font = Font(bold=True)

    # Adjust column widths
    for column in ws.columns:
        length = max(len(str(cell.value)) if cell.value is not None else 0 for cell in column)
        ws.column_dimensions[column[0].column_letter].width = length + 2

# Add title to Summary sheet
summary_ws = wb["Summary"]
summary_ws.insert_rows(1)
summary_ws["A1"] = "Telco Customer Churn - Project Summary"
summary_ws["A1"].font = Font(bold=True, size=14)

wb.save("Project_Report.xlsx")
```
